### Assignment :
Split the Company_name and Model_name from the name column.

```python
num = ['Kilometers_Driven','Mileage_value','Engine_value','Power_value']
filled = ['Price', 'Fuel_Type','Company_name', 'Model_name']
```

```python
cat = []
for i in df.columns:
    if i not in num and i not in filled:
        cat.append(i)
cat
```

# filling null values

```python
for i in cat:
    df[i].fillna(df[i].mode()[0],inplace = True)
```

```python
df.isnull().sum()
```

```python
# now for fillig the missing values in numerical columns we can use mean or median depending on outliers
```

# multivariate

```python
loca_fuel = df.groupby(["Location","Fuel_Type"])["Price"].sum().reset_index()
loca_fuel
```

```python
loca_fuel.head()
```

```python
plt.figure(figsize = (10,8))
sns.barplot(data = loca_fuel,x = 'Location',y = 'Price',hue = 'Fuel_Type')
plt.xticks(rotation = 45)
plt.show()
```

```python
local_fuel = df.groupby(["Transmission","Fuel_Type"])["Price"].mean().reset_index()
local_fuel
```

```python
#fuel type and transmission wise average sales
sns.barplot(data = local_fuel , x = 'Fuel_Type',y = 'Price',hue = 'Transmission')
plt.show()
```

```python
sns.scatterplot(data = local_fuel,x = "Fuel_Type",y = "Price",hue = "Transmission")
plt.show()
```

```python
#comany name and transmission wise average price
com_tr = df.groupby(["Company_name","Transmission"])["Price"].mean().reset_index()
sns.scatterplot(data =com_tr  ,x = "Company_name",y = "Price",hue = "Transmission")
plt.xticks(rotation = 90)
plt.grid(axis ='x')
plt.show()

```
